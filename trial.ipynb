```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import lightkurve as lk
import pandas as pd
from lightkurve import LightCurve

USE_NUMBER = 1

search_result = lk.search_lightcurve("TIC 219107776", author="SPOC", mission="TESS")
lcs = search_result.download_all()
lc : LightCurve = lcs[:USE_NUMBER].stitch(corrector_func=None)
lcf = lc.copy(copy_data=True)
```

```python
detrended_lc, trend_removed = lc.flatten(window_length=3000, return_trend=True)
```

```python
normalized_lc = lcf.normalize()
```

```python
plt.figure(figsize=(12, 10))

# Create subplots
plt.subplot(2, 1, 1)
lc.plot(ax=plt.gca(), label='Original LC', color='blue')
plt.title('Original Light Curve')
plt.legend()

plt.subplot(2, 1, 2)
normalized_lc.plot(ax=plt.gca(), label='Normalized LC', color='orange')
plt.title('Normalized Light Curve')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

lc.plot(ax=plt.gca(), label='Original LC', color='blue')
detrended_lc.plot(ax=plt.gca(), label='Detrended LC', color='red')
trend_removed.plot(ax=plt.gca(), label='Trend Removed', color='green')

plt.title('Light Curve Comparison')
plt.legend()
plt.tight_layout()
plt.show()
```
